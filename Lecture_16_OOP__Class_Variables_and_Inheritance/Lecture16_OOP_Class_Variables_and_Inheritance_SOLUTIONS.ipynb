Here is the `Student` class from Lecture 15, unchanged.

````python
class Student:
    """One student in DATASCI 160 who is looking for a lab partner."""

    def __init__(self, name, lab_skills, strategy = 0):
        self.name = name
        self.lab_skills = lab_skills
        self.strategy = strategy
        self.benchmark = 0
        self.num_met = 0
        self.partner = None

    def __str__(self):
        return f"{self.name} ({self.lab_skills})"

    # ---------- getters ----------

    def get_name(self):
        return self.name

    def get_lab_skills(self):
        return self.lab_skills

    def get_strategy(self):
        return self.strategy

    def get_benchmark(self):
        return self.benchmark

    def get_num_met(self):
        return self.num_met

    def get_partner(self):
        return self.partner

    # ---------- setters ----------

    def set_strategy(self, N):
        assert N >= 0, "a strategy cannot be negative"
        self.strategy = N

    # ---------- deciding ----------

    def rate(self, other):
        """How much do I want this person as a partner? Higher is better."""
        return other.get_lab_skills()

    def consider(self, other):
        """Meet one person. Return True if I would take them as my partner."""
        self.num_met = self.num_met + 1

        if self.num_met <= self.strategy:            # still watching
            if self.rate(other) > self.benchmark:
                self.benchmark = self.rate(other)
            return False

        return self.rate(other) > self.benchmark     # deciding

    def pair_with(self, other):
        """Make self and other each other's partner."""
        self.partner = other
        other.partner = self
````

### Some Functions that we used in Lecture 15

````python
def build_roster(N):
    """A fresh list of Student objects, one per entry in lab_partners, all using strategy N."""
    roster = []
    for name in list(lab_partners.keys()):
        skill = lab_partners[name]["lab_skills"]
        roster.append(Student(name, skill, N))
    return roster

def run_session(students, is_last_session):
    """One lab session. Everybody still looking is seated in a random pair."""

    # (1) who is still looking?
    looking = []
    for person in students:
        if person.get_partner() == None:
            looking.append(person)

    # (2) seat them in random pairs
    random.shuffle(looking)

    # (3) each pair decides
    for i in range(0, len(looking)//2):
        a = looking[i]
        b = looking[-i]
        a_says_yes = a.consider(b)       # ask both, so that both of them
        b_says_yes = b.consider(a)       # count the meeting

        if a_says_yes and b_says_yes:
            a.pair_with(b)
        elif is_last_session:            # no sessions left: take whoever you are sitting with
            a.pair_with(b)


def run_semester(students, num_sessions):
    """A whole semester of lab sessions. The last session is the deadline."""
    for session in range(1, num_sessions + 1):
        is_last_session = (session == num_sessions)
        run_session(students, is_last_session)
````

## <span style="color:red"> II. Composition: an object that holds objects </span>

<font size = "4">

A `LabSection` stores a list of `Student` objects as one of its attributes. An object whose
attributes include other objects is called **composition**. 

````python
class LabSection:
    """One lab section of DATASCI 160: its students and the semester they go through."""

    def __init__(self, partner_dict, strategy):
        self.students = []
        for name in partner_dict.keys():
            skill = partner_dict[name]["lab_skills"]
            self.students.append(Student(name, skill, strategy))

    def __str__(self):
        return f"A lab section of {len(self.students)} students"

    def get_student(self, name):
        """The Student with this name, or None if there is nobody by that name.""" #*** NEW
        for person in self.students:
            if person.get_name() == name:
                return person
        return None

    def run_session(self, is_last_session):
        """One lab session. Everybody still looking is seated in a random pair."""
        looking = []
        for person in self.students:
            if person.get_partner() == None:
                looking.append(person)

        random.shuffle(looking)

        for i in range(0, len(looking), 2):
            a = looking[i]
            b = looking[i + 1]

            a_says_yes = a.consider(b)
            b_says_yes = b.consider(a)

            if a_says_yes and b_says_yes:
                a.pair_with(b)
            elif is_last_session:
                a.pair_with(b)

    def run_semester(self, num_sessions):
        """A whole semester. The last session is the deadline."""
        for session in range(1, num_sessions + 1):
            is_last_session = (session == num_sessions)
            self.run_session(is_last_session)
````

## <span style="color:red"> III. A class variable </span>

<font size = "4">

Let's change `num_sessions` into a class variable.

````python
class LabSection:
    """One lab section of DATASCI 160: its students and the semester they go through."""

    num_sessions = 12 # every lab section meets 12 times in a semester

    def __init__(self, partner_dict, strategy):
        self.students = []
        for name in list(partner_dict.keys()):
            skill = partner_dict[name]["lab_skills"]
            self.students.append(Student(name, skill, strategy))

    def __str__(self):
        return f"A lab section of {len(self.students)} students"

    def get_student(self, name):
        """The Student with this name, or None if there is nobody by that name."""
        for person in self.students:
            if person.get_name() == name:
                return person
        return None

    def run_session(self, is_last_session):
        """One lab session. Everybody still looking is seated in a random pair."""
        looking = []
        for person in self.students:
            if person.get_partner() == None:
                looking.append(person)

        random.shuffle(looking)

        for i in range(0, len(looking), 2):
            a = looking[i]
            b = looking[i + 1]

            a_says_yes = a.consider(b)
            b_says_yes = b.consider(a)

            if a_says_yes and b_says_yes:
                a.pair_with(b)
            elif is_last_session:
                a.pair_with(b)

    def run_semester(self):
        """A whole semester. The last session is the deadline.""" # uses the class variable to know how many sessions in a semester
        for session in range(1, LabSection.num_sessions + 1):
            is_last_session = (session == LabSection.num_sessions)
            self.run_session(is_last_session)
````

First we'll modify the strategy setter for this new type of strategy:

````python
class Student:
    """One student in DATASCI 160 who is looking for a lab partner."""

    def __init__(self, name, lab_skills, strategy = 0):
        self.name = name
        self.lab_skills = lab_skills
        self.strategy = strategy
        self.benchmark = 0
        self.num_met = 0
        self.partner = None

    def __str__(self):
        return f"{self.name} ({self.lab_skills})"

    # ---------- getters ----------

    def get_name(self):
        return self.name

    def get_lab_skills(self):
        return self.lab_skills

    def get_strategy(self):
        return self.strategy

    def get_benchmark(self):
        return self.benchmark

    def get_num_met(self):
        return self.num_met

    def get_partner(self):
        return self.partner

    # ---------- setters ----------

    def set_strategy(self, N, *, floor = False):
        assert N >= 0, "a strategy cannot be negative"
        self.strategy = N
        if floor == True:
            self.strategy = 0
            self.benchmark = N

    # Let's also make a lab skills setter that we'll use in a minute
    def set_skill(self, skill_level):
        self.lab_skills = skill_level

    # ---------- deciding ----------

    def rate(self, other):
        """How much do I want this person as a partner? Higher is better."""
        return other.get_lab_skills()

    def consider(self, other):
        """Meet one person. Return True if I would take them as my partner."""
        self.num_met = self.num_met + 1

        if self.num_met <= self.strategy:            # still watching
            if self.rate(other) > self.benchmark:
                self.benchmark = self.rate(other)
            return False

        return self.rate(other) > self.benchmark     # deciding

    def pair_with(self, other):
        """Make self and other each other's partner."""
        self.partner = other
        other.partner = self
````

### <span style="color:red"> A systematic experiment </span>

<font size = "4">

How to approach the question systematically:

- First of all, we should be using randomly selected lab skills instead of the same set of lab skills every time.
- Let's pick out the first ten students, and assign them skills = 0.5, 1.5, 2.5, ..., 9.5 for testing
  - These students will use the new floor strategy, and we'll check which floor is optimal for each skill level
  - But, if we do this, we should move to 100 students rather than 50 so that our test students aren't 20% of the class


````python
greendale_students = pd.read_csv("data/greendale_students.csv")

names = list(greendale_students["name"])
test_names = names[0:10]

print(len(names), "names")
print("test students:", test_names)
````

Here's a quick function to create a dictionary with randomly assigned lab skills from our list of names:

````python
def random_class():
    class_dict = {}
    for i in range(len(names)):
        if i < 10:
            skill = i + 0.5
        else:
            skill = round(random.uniform(0, 10), 2)
        class_dict[names[i]] = {"lab_skills": skill}
    return class_dict


section = LabSection(random_class(), 3)
print(section)
for name in test_names:
    print(section.get_student(name))
````

#### <span style="color:red"> For each of our test students, let's see what the best floor is</span>

First we'll define a function for testing one of our test students with a given floor

````python
def test_with_floor(n, f):
    n_partner_quality = []

    for _ in range(200):

        section = LabSection(random_class(), 3)

        n_student = section.get_student(n)
        n_student.set_strategy(f, floor = True)

        section.run_semester()
        n_partner_quality.append(n_student.get_partner().get_lab_skills())
    return np.mean(n_partner_quality)
````

Now we'll do a quick loop over our test students and possible floors:

````python
for person in test_names:
    best_floor = 0
    mean_partner = 0
    for f in range(1, 10):
        mp = test_with_floor(person, f)
        if mp > mean_partner:
            best_floor = f
            mean_partner = mp
    print(f"{section.get_student(person)} had {best_floor} as their best floor, giving them on average a partner with {mean_partner} lab skills")
````

## <span style="color:red"> V. Inheritance: a study group </span>

The **child** class starts with everything the **parent** class has.

````python
class StudyGroupStudent(Student):
    """A student who is also in the study group."""
    pass
````

### <span style="color:red"> Overriding a method </span>

<font size = "4">

While `StudyGroupStudent` inherits all of the methods that a `Student` has, we can overide those methods to be something else:

````python
class StudyGroupStudent(Student):
    """A student who is also in the study group."""

    def __str__(self):
        return f"{self.name} ({self.lab_skills}, study group)"
````

````python
abed = StudyGroupStudent("Abed", 3.24)
todd = Student("Todd", 9.99)

print(abed)       # the child's __str__
print(todd)       # the parent's __str__
````

### <span style="color:red"> Being realistic in your critical thinking </span>

<font size = "4">

Earlier we asked: What is the best strategy if everyone else is using the interview 3 for benchmark strategy?

Now we ask: **What is the best strategy if some people are acting irrationally?**


Specifically, let's imagine that some of the students in the class are friends through a study group. They might rate each other by applying a bonus on top of the other study group member's lab skills.

````python
class StudyGroupStudent(Student):
    """A student who is also in the study group."""

    bonus = 5 # notice that this is a class variable, just like we used in LabSection
    def __str__(self):
        return f"{self.name} ({self.lab_skills}, study group)"

    def rate(self, other):
        """Fellow members get a bonus on top of their lab skills."""
        if isinstance(other, StudyGroupStudent):
            return other.get_lab_skills() + StudyGroupStudent.bonus
        return other.get_lab_skills()
````

````python
abed = StudyGroupStudent("Abed", 3.24)
troy = StudyGroupStudent("Troy", 5.66)
todd = Student("Todd", 9.99)

print("Abed rates Troy:", abed.rate(troy))
print("Abed rates Todd:", abed.rate(todd))
print("Todd rates Troy:", todd.rate(troy))
````

<font size = "4">

Note that the `consider` method that `StudyGroupStudent` inherits from `Student` immediately works with the new `rate` method: 

````python
abed = StudyGroupStudent("Abed", 3.24, 1)
natalie = Student("Natalie", 8.16)
troy = StudyGroupStudent("Troy", 5.66)

print("meets Natalie:", abed.consider(natalie), "| benchmark", abed.get_benchmark())
print("meets Troy   :", abed.consider(troy))
````

## <span style="color:red"> VI. A class method </span>

<font size = "4">

A **class method** allows us to change a class variable for all instances. We already did this with `LabSection.num_sessions`. However, if this is something you expect users to manipulate, it is best practice to define a setter so that we can do things like use assert to ensure appropriate values for the bonus.

````python
class StudyGroupStudent(Student):
    """A student who is also in the study group."""

    bonus = 5

    def __str__(self):
        return f"{self.name} ({self.lab_skills}, study group)"

    def rate(self, other):
        """Fellow members get a bonus on top of their lab skills."""
        if isinstance(other, StudyGroupStudent):
            return other.get_lab_skills() + StudyGroupStudent.bonus
        return other.get_lab_skills()

    @classmethod
    def set_bonus(cls, value):
        """Change the bonus for every member of the study group at once."""
        assert value >= 0, "the bonus cannot be negative"
        cls.bonus = value
````

````python
abed = StudyGroupStudent("Abed", 3.24)
troy = StudyGroupStudent("Troy", 3.66)

print("bonus 5 :", abed.rate(troy))

StudyGroupStudent.set_bonus(7)
print("bonus 7:", abed.rate(troy)) # Abed was initialized before the change, but still changes

StudyGroupStudent.set_bonus(-1)
````
