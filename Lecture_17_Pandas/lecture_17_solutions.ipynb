````python
# set up code for next exercise

df_ht_wt = pd.read_csv("data_raw/SOCR-HeightWeight.csv")

# Grab height and weight and convert to correct units

def pounds_to_kg(wt):
    return wt*0.45359237

def inches_to_meters(ht):
    ht_mm = ht*25.4
    return ht_mm*1e-3

ht_inch = df_ht_wt["Height(Inches)"]
wt_lbs = df_ht_wt["Weight(Pounds)"]

ht_meters = ht_inch.apply(inches_to_meters)
ht_meters.name = "Height(Meters)"
wt_kg = wt_lbs.apply(pounds_to_kg)
wt_kg.name = "Weight(Kilograms)"
````

**Exercise:** Compute a histogram of "mpg" by cylinder count for cars with 4, 6, and 8 cylinders.Here is what it might look like (using a fake dataset):

<img src="data_raw/fake_histogram.png" alt="drawing" width="400"/>


You can make the histograms transparent by adjusting ```alpha``` in

```python
    plt.hist(x = ..., alpha = 0.5)
```

````python
import matplotlib.pyplot as plt

cylinders = [4, 6, 8]

for c in cylinders:
    df = carfeatures.query("cylinders == @c")
    plt.hist(x=df["mpg"], alpha=0.5)

# Add these AFTER the loop
plt.legend(labels = cylinders, title = "# of cylinders")   
plt.xlabel("MPG")
plt.ylabel("Number of Cars")
plt.title("Histogram of MPG by Cylinder Count")
plt.show()
````

**Exercise (social):** Loop over the unique values of `Primary_Platform` and make a scatter plot of `Daily_Usage_Hours` vs. `Sleep_Duration_Hours`, one color per platform. Add axis labels and a legend.

````python
for platform in social["Primary_Platform"].unique():
    df = social.query("Primary_Platform == @platform")
    plt.scatter(df["Daily_Usage_Hours"], df["Sleep_Duration_Hours"])

plt.legend(labels = social["Primary_Platform"].unique(), title = "Primary Platform")   
plt.xlabel("Daily Usage (Hours)")
plt.ylabel("Sleep Duration (Hours)")
plt.show()
````

**Exercise (coffee):** Use `pd.unique()` to find the categories of `order_method`. Then use a loop to make a scatter plot of `discount_voucher_pct` vs. `daily_cups_sold`, one color per order method, with a legend.

````python
categories = pd.unique(coffee["order_method"])

for cat in categories:
    df = coffee.query("order_method == @cat")
    plt.scatter(df["discount_voucher_pct"], df["daily_cups_sold"])

plt.legend(labels = categories, title = "Order Method")
plt.xlabel("Discount Voucher %")
plt.ylabel("Daily Cups Sold")
plt.show()
````
