## <span style="color:red"> II. A Student class </span>

<font size = "4">

Think back to `run_one_trial(N)`. While its loop ran, a couple of local variables kept track
of where you were: the benchmark so far, and how far down the line you had got. When the
function returned, they were forgotten. That was fine, because only one person was
searching.

Now 50 people are searching at once, and each one is at a different point in their own
search. Todd might have met four people and have a benchmark of 8.16, while Craig has met
two and has a benchmark of 1.44. Each student needs their own copy of those variables,
and the copies need to last all semester.

That is what an object gives us. Each `Student` object will store:

| attribute | what it means | changes during the semester? |
|---|---|---|
| `name` | the student's name | no |
| `lab_skills` | their lab skills, out of 10 | no |
| `strategy` | N: how many people they watch before deciding | no |
| `benchmark` | the best lab skills they have seen while watching | **yes** |
| `num_met` | how many people they have met so far | **yes** |
| `partner` | the student they end up with | **yes** |



```python
class Student:
    """One student in DATASCI 160 who is looking for a lab partner."""

    def __init__(self, name, lab_skills, strategy = 0):
        self.name = name
        self.lab_skills = lab_skills
        self.strategy = strategy
        self.benchmark = 0
        self.num_met = 0
        self.partner = None

    def __str__(self):
        return f"{self.name} ({self.lab_skills})"
```

<font size = "4">

- `__init__` has three parameters, and `strategy` has a **default** of 0.
- `benchmark`, `num_met` and `partner` are not parameters at all. Every student starts the
  semester the same way: no benchmark yet, nobody met, no partner.


```python
todd = Student("Todd", 9.99, 2)
craig = Student("Craig", 0.37)          # strategy takes its default, 0

print(todd)
print(craig)
print(todd.strategy, craig.strategy)
print(todd.benchmark, todd.num_met, todd.partner)
```

<font size = "4">

One class, two objects, each with its own attributes: two apartments built from the same
floor plan.

## <span style="color:red"> III. Getters and setters </span>

<font size = "4">

So far we have reached into objects directly, as in `todd.strategy`. That works, but nothing
stops a mistake:

```python
todd.strategy = -3        # doesn't make sense as a strategy
print(todd.strategy)
```

<font size = "4">

A negative strategy makes no sense, and nothing complained. The usual fix is to

- read attributes through **getter** methods (`get_name()`, `get_benchmark()`, ...), and
- change them through **setter** methods (`set_strategy(N)`).

A setter is the one place a value can change, so it is the natural place to check it, with
an `assert` like the one you added to `Fraction.__init__` in Lecture 14.

```python
class Student:
    """One student in DATASCI 160 who is looking for a lab partner."""

    def __init__(self, name, lab_skills, strategy = 0):
        self.name = name
        self.lab_skills = lab_skills
        self.strategy = strategy
        self.benchmark = 0
        self.num_met = 0
        self.partner = None

    def __str__(self):
        return f"{self.name} ({self.lab_skills})"

    # ---------- getters ----------

    def get_name(self):
        return self.name

    def get_lab_skills(self):
        return self.lab_skills

    def get_strategy(self):
        return self.strategy

    def get_benchmark(self):
        return self.benchmark

    def get_num_met(self):
        return self.num_met

    def get_partner(self):
        return self.partner

    # ---------- setters ----------

    def set_strategy(self, N):
        assert N >= 0, "a strategy cannot be negative"
        self.strategy = N
```

```python
todd = Student("Todd", 9.99)
print(todd.get_name(), todd.get_lab_skills(), todd.get_strategy())

todd.set_strategy(2)
print(todd.get_strategy())
```

```python

todd.set_strategy(-3)
```

<font size = "4">

The setter refused the bad value on the line where the mistake was made.

From now on, code outside the class uses getters and setters. Code inside the class,
in its own methods, uses `self.benchmark` and friends directly.

## <span style="color:red"> IV. The method that decides </span>

<font size = "4">

Now the method that does the real work. `todd.consider(other)` means: *Todd has just met
`other`. Would Todd take them as a partner?* It returns `True` or `False`.

It is the benchmark rule from Assignment 4, written from inside one student:

1. Count the meeting.
2. If I have met N people or fewer, I am **still watching**. If this person beats my
   benchmark, remember their lab skills. Either way, say no.
3. Otherwise I am **deciding**: say yes if this person beats my benchmark.

We also add a small method, `rate(other)`, that answers *"how much do I want this
person?"* For now the answer is just their lab skills, so `rate` looks pointless. We give
that question its own method because next lecture some students will answer it differently.

```python
class Student:
    """One student in DATASCI 160 who is looking for a lab partner."""

    def __init__(self, name, lab_skills, strategy = 0):
        self.name = name
        self.lab_skills = lab_skills
        self.strategy = strategy
        self.benchmark = 0
        self.num_met = 0
        self.partner = None

    def __str__(self):
        return f"{self.name} ({self.lab_skills})"

    # ---------- getters ----------

    def get_name(self):
        return self.name

    def get_lab_skills(self):
        return self.lab_skills

    def get_strategy(self):
        return self.strategy

    def get_benchmark(self):
        return self.benchmark

    def get_num_met(self):
        return self.num_met

    def get_partner(self):
        return self.partner

    # ---------- setters ----------

    def set_strategy(self, N):
        assert N >= 0, "a strategy cannot be negative"
        self.strategy = N

    # ---------- deciding ----------

    def rate(self, other):
        """How much do I want this person as a partner? Higher is better."""
        return other.get_lab_skills()

    def consider(self, other):
        """Meet one person. Return True if I would take them as my partner."""
        self.num_met = self.num_met + 1

        if self.num_met <= self.strategy:            # still watching
            if self.rate(other) > self.benchmark:
                self.benchmark = self.rate(other)
            return False

        return self.rate(other) > self.benchmark     # deciding
```

<font size = "4">

Let's trace it. Todd uses N = 2: he watches two people, then decides.

```python
todd = Student("Todd", 9.99, 2)
craig = Student("Craig", 0.37)
duncan = Student("Duncan", 9.76)
chang = Student("Chang", 5.07)

print("meets Craig :", todd.consider(craig), "| benchmark", todd.get_benchmark(), "| met", todd.get_num_met())
print("meets Duncan:", todd.consider(duncan), "| benchmark", todd.get_benchmark(), "| met", todd.get_num_met())
print("meets Chang :", todd.consider(chang), "| benchmark", todd.get_benchmark(), "| met", todd.get_num_met())
```

<font size = "4">

Todd said no three times, for two different reasons. Craig and Duncan were turned down
because Todd was still watching. Chang was turned down because 5.07 does not beat 9.76.

Every call changed Todd: `num_met` went up each time, and `benchmark` went up twice. **Todd
remembers his meetings.** In `run_one_trial` the loop did the remembering for us; now the
object does it.

Look at Todd's benchmark: 9.76. The only student in the class with lab skills above 9.76 is
Todd himself. Unless the deadline forces it, Todd will not say yes to anybody this semester.
Keep that in mind.

Now the other half of a meeting:

```python
craig = Student("Craig", 0.37)          # strategy 0: decides straight away
todd = Student("Todd", 9.99, 2)

craig_says_yes = craig.consider(todd)
todd_says_yes = todd.consider(craig)

print("Craig says   :", craig_says_yes)
print("Todd says    :", todd_says_yes)
print("partnership? :", craig_says_yes and todd_says_yes)
```

<font size = "4">

Craig would happily take Todd. It doesn't matter: a partnership needs both.

## <span style="color:red"> V. Objects that point at objects </span>

<font size = "4">

When two students agree, each one needs to remember *who* their partner is. A partner is
not a name or a number. It is the other `Student` object itself.

One method sets up both sides at once:

```python
class Student:
    """One student in DATASCI 160 who is looking for a lab partner."""

    def __init__(self, name, lab_skills, strategy = 0):
        self.name = name
        self.lab_skills = lab_skills
        self.strategy = strategy
        self.benchmark = 0
        self.num_met = 0
        self.partner = None

    def __str__(self):
        return f"{self.name} ({self.lab_skills})"

    # ---------- getters ----------

    def get_name(self):
        return self.name

    def get_lab_skills(self):
        return self.lab_skills

    def get_strategy(self):
        return self.strategy

    def get_benchmark(self):
        return self.benchmark

    def get_num_met(self):
        return self.num_met

    def get_partner(self):
        return self.partner

    # ---------- setters ----------

    def set_strategy(self, N):
        assert N >= 0, "a strategy cannot be negative"
        self.strategy = N

    # ---------- deciding ----------

    def rate(self, other):
        """How much do I want this person as a partner? Higher is better."""
        return other.get_lab_skills()

    def consider(self, other):
        """Meet one person. Return True if I would take them as my partner."""
        self.num_met = self.num_met + 1

        if self.num_met <= self.strategy:            # still watching
            if self.rate(other) > self.benchmark:
                self.benchmark = self.rate(other)
            return False

        return self.rate(other) > self.benchmark     # deciding

    def pair_with(self, other):
        """Make self and other each other's partner."""
        self.partner = other
        other.partner = self
```

```python
todd = Student("Todd", 9.99, 2)
duncan = Student("Duncan", 9.76, 2)
craig = Student("Craig", 0.37)

todd.pair_with(duncan)

print(todd.get_partner())
print(duncan.get_partner())
print(craig.get_partner())
```

<font size = "4">

`pair_with` changed two objects: the one it was called on (`self`) and the one passed in
(`other`).

Because a partner is a whole `Student`, you can keep following the dots. Read each chain
left to right: `todd.get_partner()` is the Duncan object, so the next method is called on
Duncan.

```python
print(todd.get_partner().get_name())
print(todd.get_partner().get_lab_skills())
print(todd.get_partner().get_partner())       # Todd's partner's partner is... Todd
```

<font size = "4">

Craig has no partner. What happens if we follow the dots anyway?

```python

print(craig.get_partner().get_name())
```

<font size = "4">

`craig.get_partner()` is `None`, and `None` has no `get_name` method. Whenever an error
mentions `'NoneType'`, ask yourself which object you expected to exist and didn't.

## <span style="color:red"> VI. A whole class of students </span>

<font size = "4">

The same loop we used all through Assignment 4 turns the `lab_partners` dictionary into a
list of `Student` objects:

```python
def build_roster(N):
    """A fresh list of Student objects, one per entry in lab_partners, all using strategy N."""
    roster = []
    for name in list(lab_partners.keys()):
        skill = lab_partners[name]["lab_skills"]
        roster.append(Student(name, skill, N))
    return roster


roster = build_roster(2)

print(len(roster), "students")
print(roster[0])
print(roster[12])
```

<font size = "4">

`roster` is an ordinary list. `len`, indexing, `random.shuffle` and `for` loops all work as
before. Only the contents are new.

Printing the whole list is less friendly:

```python
print(roster[0:3])
```

<font size = "4">

Inside a list, Python prints each object with `__repr__`, not `__str__`, and we haven't
written a `__repr__`. (Exercise 1 of the homework fixes this.)

A running maximum works on objects too:

```python
best = roster[0]

for person in roster:
    if person.get_lab_skills() > best.get_lab_skills():
        best = person

print("The strongest student is", best)
```

<font size = "4">

## <span style="color:red"> VII. A lab session and a semester </span>

<font size = "4">

One lab session has three steps:

1. Make a list of everybody who is still looking (their partner is `None`).
2. Shuffle it, so the seating is random.
3. Go down the list two at a time. Each pair considers each other, and if both say yes,
   they pair up. At the last session, they pair up regardless.


<font size = "4">

There is always an even number of people looking: everybody starts out looking, and people
stop looking two at a time. So the list always splits evenly into pairs.

```python
def run_session(students, is_last_session):
    """One lab session. Everybody still looking is seated in a random pair."""

    # (1) who is still looking?
    looking = []
    for person in students:
        if person.get_partner() == None:
            looking.append(person)

    # (2) seat them in random pairs
    random.shuffle(looking)

    # (3) each pair decides
    for i in range(0, len(looking), 2):
        a = looking[i]
        b = looking[i + 1]

        a_says_yes = a.consider(b)       # ask both, so that both of them
        b_says_yes = b.consider(a)       # count the meeting

        if a_says_yes and b_says_yes:
            a.pair_with(b)
        elif is_last_session:            # no sessions left: take whoever you are sitting with
            a.pair_with(b)
```

<font size = "4">

Two things to notice:

- Both students are asked, on separate lines, before we decide anything. That way both of
  them count the meeting, even when the first one says no.
- The deadline lives here, not in `consider`. `consider` is about what one student wants.
  The deadline is a rule of the course, so it belongs to the session.

Try one session where everybody uses N = 0:

```python
roster = build_roster(0)
run_session(roster, False)

count = 0
for person in roster:
    if person.get_partner() == None:
        count = count + 1

print(count, "students still looking after one session")
```

<font size = "4">

With N = 0 nobody watches, the benchmark stays at 0, and every lab skill beats 0, so
everybody says yes to the first person they meet. The whole class pairs off in session 1,
completely at random.

Now everybody uses N = 2:

```python
roster = build_roster(2)
run_session(roster, False)

count = 0
for person in roster:
    if person.get_partner() == None:
        count = count + 1

print(count, "students still looking after one session")
```

<font size = "4">

Nobody pairs up, because everybody is still watching. With N = 2, the first partnership
cannot form before session 3.

A semester is just 12 sessions in a row, with the last one marked as the deadline:

```python
def run_semester(students, num_sessions):
    """A whole semester of lab sessions. The last session is the deadline."""
    for session in range(1, num_sessions + 1):
        is_last_session = (session == num_sessions)
        run_session(students, is_last_session)
```

```python
roster = build_roster(2)
run_semester(roster, 12)

for position in range(5):
    person = roster[position]
    print(person, "ended up with", person.get_partner())
```

<font size = "4">

Everybody ends up with somebody, because the last session forces it.

## <span style="color:red"> VIII. Mutation: objects change </span>

<font size = "4">

In Lecture 14 we met two kinds of method:

- `f1 + f2` called `__add__`, which built and returned a new `Fraction`. `f1` and `f2`
  were unchanged.
- `my_fraction.reduce()` changed `my_fraction` itself and returned nothing.

Changing an object in place is called **mutation**. `Student` has both kinds:

| method | changes the student? | returns |
|---|---|---|
| getters, `rate` | no | a value |
| `set_strategy` | yes | nothing |
| `consider` | yes: `num_met`, and sometimes `benchmark` | `True` or `False` |
| `pair_with` | yes: **two** students | nothing |

```python
roster = build_roster(2)
todd = roster[12]

print("before:", todd.get_benchmark(), todd.get_num_met(), todd.get_partner())

run_semester(roster, 12)

print("after :", todd.get_benchmark(), todd.get_num_met(), todd.get_partner())
```

<font size = "4">

We never mentioned `todd` in the line that ran the semester, and yet Todd changed. `todd`
and `roster[12]` are two names for the **same** object, so when `run_semester` changed
`roster[12]`, it changed Todd.

What happens if we run a second semester on the same roster?

```python
run_semester(roster, 12)

print("after a second semester:", todd.get_benchmark(), todd.get_num_met(), todd.get_partner())
```

<font size = "4">

Nothing. Everybody already has a partner, so nobody is looking and every session is empty.
**A roster that has been through a semester is used up.** It carries every benchmark, every
count and every partnership from that semester.

So we build a **fresh roster for every semester**, the same way `run_one_trial` shuffled a
fresh interview order for every trial:

```python
for semester in range(5):
    roster = build_roster(2)
    run_semester(roster, 12)
    print("semester", semester, ": Todd ended up with", roster[12].get_partner())
```

```python
# your code here

class Student:
    """One student in DATASCI 160 who is looking for a lab partner."""

    def __init__(self, name, lab_skills, strategy = 0):
        self.name = name
        self.lab_skills = lab_skills
        self.strategy = strategy
        self.benchmark = 0
        self.num_met = 0
        self.partner = None

    def __str__(self):
        return f"{self.name} ({self.lab_skills})"

    # ---------- getters ----------

    def get_name(self):
        return self.name

    def get_lab_skills(self):
        return self.lab_skills

    def get_strategy(self):
        return self.strategy

    def get_benchmark(self):
        return self.benchmark

    def get_num_met(self):
        return self.num_met

    def get_partner(self):
        return self.partner

    # ---------- setters ----------

    def set_strategy(self, N):
        assert N >= 0, "a strategy cannot be negative"
        self.strategy = N

    # ---------- deciding ----------

    def rate(self, other):
        """How much do I want this person as a partner? Higher is better."""
        return other.get_lab_skills()

    def consider(self, other):
        """Meet one person. Return True if I would take them as my partner."""
        self.num_met = self.num_met + 1

        if self.num_met <= self.strategy:            # still watching
            if self.rate(other) > self.benchmark:
                self.benchmark = self.rate(other)
            return False

        return self.rate(other) > self.benchmark     # deciding

    def pair_with(self, other):
        """Make self and other each other's partner."""
        self.partner = other
        other.partner = self

    # **************************************************
    # ********** SOLUTION ******************************
    # **************************************************

    def report(self):
        if self.get_partner() != None:
            part_rep = self.get_partner().get_name()
        else:
            part_rep = "nobody"
        print(f"{self.get_name()} | lab skills {self.get_lab_skills()} | strategy {self.get_strategy()} | "
              f"benchmark {self.get_benchmark()} | met {self.get_num_met()} | partner {part_rep}")
```

**4.** Suppose `pair_with` had been written with only its first line, `self.partner = other`.
The cell below imitates that by setting just one side. (It reaches into the attribute
directly, which we would not normally do.)

- What does the cell print?
- Y still looks like it is searching. Explain in a sentence or two what would go wrong at the

```python
x = Student("X", 5.0)
y = Student("Y", 7.0)

x.partner = y          # only the first line of pair_with

print(x.get_partner(), "|", y.get_partner())
```

Answer:

The cell prints: Y (7.0) | None

What goes wrong is that x is assigned y as a partner, but y will still be treated as not having a partner. This can likely cause a crash because our indexing over the students who are still looking assumes we have an even number, but if only one person in a pair is actually assigned a partner we can end up with an odd number of students looking and looking[i + 1] will give an index error.
