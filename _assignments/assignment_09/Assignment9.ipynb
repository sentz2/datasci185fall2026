A weather station logged one million readings of temperature (°C) and relative humidity (%). **Negative humidity readings are sensor errors.**

In the setup cell, we write a loop-based function that computes the following:

1. `n_errors`: the number of readings with negative humidity
2. `n_hot_dry`: the number of readings with temperature above 30 **and** humidity from 0 to 30 (both inclusive)
3. `mean_temp_extreme`: the mean temperature of readings that are below 10 degrees Celsius **or** above 35 degrees Celsius
4. `humidity_clean`: an array that has the same values as `humidity` but every negative value is replaced with `np.nan`
5. `mean_humidity`: the mean of `humidity_clean`, ignoring the missing values. 


```python
np.random.seed(160)
n_readings = 1_000_000
temp_c = np.random.normal(22, 8, n_readings).round(1)
humidity = np.random.uniform(-5, 100, n_readings).round(1)


def loop_solution():
    # Part 1
    n_errors = 0
    for h in humidity:
        if h < 0:
            n_errors += 1

    # Part 2
    n_hot_dry = 0
    for t, h in zip(temp_c, humidity):
        if (t > 30) and (0 <= h <= 30):
            n_hot_dry += 1

    # Part 3
    mean_temp_extreme = 0
    mean_temp_ct = 0
    for t in temp_c:
        if (t < 10) or (t > 35):
            mean_temp_extreme += t
            mean_temp_ct += 1
    mean_temp_extreme /= mean_temp_ct

    # Part 4
    humidity_clean = humidity.copy()
    for k in range(len(humidity_clean)):
        if humidity_clean[k] < 0:
            humidity_clean[k] = np.nan

    # Part 5
    mean_humidity = 0
    mean_hum_ct = 0
    for h in humidity_clean:
        if np.isnan(h):
            continue
        mean_humidity += h
        mean_hum_ct += 1
    mean_humidity /= mean_hum_ct
    
    return n_errors, n_hot_dry, mean_temp_extreme, humidity_clean, mean_humidity

OUT = loop_solution()
print("n_errors:          ", OUT[0])
print("n_hot_dry:         ", OUT[1])
print("mean_temp_extreme: ", OUT[2])
print("humidity_clean[:10]", OUT[3][:10])
print("mean_humidity:     ", OUT[4])
```

A stick of length 1 is broken at two random points, giving three pieces. Can the three pieces form a triangle? They can if each pair of pieces is longer, together, than the third piece.

The cell below makes one million pairs of random cut points (`cut1` and `cut2` are NumPy arrays) and a loop-based function, `triangle_loop`, that returns the fraction of trials where the pieces form a triangle.


```python
np.random.seed(160)
n_trials = 1_000_000
cut1 = np.random.uniform(size = n_trials)
cut2 = np.random.uniform(size = n_trials)

def triangle_loop(cut1, cut2):
    count = 0
    for i in range(len(cut1)):
        left_cut = min(cut1[i], cut2[i])
        right_cut = max(cut1[i], cut2[i])
        length_1 = left_cut
        length_2 = right_cut - left_cut
        length_3 = 1 - right_cut
        if (length_1 + length_2 > length_3) and (length_1 + length_3 > length_2) and (length_2 + length_3 > length_1):
            count = count + 1
    return count / len(cut1)

print("Fraction that form a triangle:", triangle_loop(cut1, cut2))
```
